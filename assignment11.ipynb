```python
class Product:
    def __init__(self, id, name, price, qty):
        self.id = id
        self.name = name
        self.price = price
        self.qty = qty

obj = Product(101, "Mobile", 15000, 2)

print("Product ID:", obj.id)
print("Product Name:", obj.name)
print("Product Price:", obj.price)
print("Product Quantity:", obj.qty)
```
