```python
num1 = int(input("Enter num1 : "))
num2 = int(input("Enter num2 : "))
choice = input("Enter choice + - / *  : ")
```

```python
def add(num1,num2):
    return num1 +  num2 
def sub(num1,num2):
    return num1 - num2
def multi(num1,num2):
    return num1*num2
def div(num1,num2):
    return num1/num2
```

```python
def main():         #funtion definition / entry point /orchestrator
    if choice == '+' :
        sum = add(num1,num2)
        print(sum)
    elif choice == '-':
        minus = sub(num1,num2)
        print(minus)
    elif choice == '*':
        multiplication = multi(num1,num2)
        print(multiplication)
    elif choice == '/' :
        division = div(num1,num2)
        print(division)
    else :
        print("Enter correct choices ")
            
main()
```
