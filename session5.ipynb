```python
#concept of oop 
# Any function define in any class is called methods and thats why we write self
#DEfine Class

class vehicle :
    company_name = "Tata" #class variable/class attribute -   because defined inside class

    #define constructor/special method/instance - it is special method which pass attribut to other defined function inside a same class 
    #self is use to fetch any arguments from constructor 
    def __init__(self,name,age):
        self.name = name # Instnace variable / Instance Attributes
        self.age = age
    
    def sound(self):
        print(f'{self.name} and {self.age}')
        return "BEEP"

```

```python
#define object - tata is object here  
tata = vehicle()
```

```python
#fetching sound() method using . operator
tata.sound()
```

```python
#define object 
tata = vehicle('Pinky',21)
```
