```python
num = int(input("Enter a num between 0 10 100 : "))
even = []
odd = []
for i in range(0,num+1):
    if i%2 == 0:
        even.append(i)
    else:
        odd.append(i)
tup1= tuple(even)
tup2 = tuple(odd)
print(tup1)
print(tup2)
```
