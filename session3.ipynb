```python
# tuple comprehension

#even= (<expression> for loop <condition>)
even =[i for i in range(0,51) if i%2==0]
odd =[i for i in range(0,51) if i%2!=0]
print(tuple(even))
print(tuple(odd))
```

```python
# Dictionary Comprehension

# Create a dictionary using dictionary comprehension
squared_dict = {x: x**2 for x in range(1, 6)}
print(squared_dict) # Output: {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}
```

```python
# Create a dictionary from two lists using dictionary comprehension
keys = ['a', 'b', 'c']
values = [1, 2, 3]

dict =  {k : v for k,v in zip(keys,values)}
dict
```

```python
# Create a dictionary with conditional logic using dictionary comprehension
dict = {x : x** 3 for x in range(1,11) if x % 2 != 0}
dict
```

```python
# Create a dictionary with nested dictionary comprehension

nest_dict = {x : {y : y** 2 for y in range(1,4)} for x in range(1,4)}
print(nest_dict)
```

```python
# In-place means: the original set itself is changed.
a = {1, 2, 3}
b = {2, 3, 4}

a.intersection_update(b)

# Now a becomes:

{2, 3}

# The original set a was modified.
```
