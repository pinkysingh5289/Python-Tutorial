```python
class parent :
    def __init__(self,name,age,
                 gender,department,salary):
        self.name = name
        self.age = age
        self.gender = gender 
        self.department = department
        self.sal = salary
    def hobbies(self):
        print(f"{self.sal}")
    
```

```python
class child1(parent):
    def __init__(self,name,age,
                 gender,department,salary,
                 height,weight): # it can have parent class attributes  and also its own attribute 
        super().__init__(name,age,
                 gender,department,salary)
        self.height = height #it can have self.name 
        self.weight = weight

    def hobbies(self):
        print
```

```python
obj1 = parent('a',60,'m','m',12345)
```

```python
from abc import ABC,abstractmethod

class father1(ABC):

    def __init__(self,name,age,gender,department,salary):
        self.name = name
        self.age = age
        self.gender = gender
        self.department =  department
        self.salary = salary

    @abstractmethod
    def hobbies(self):
        '''abstract method implement by child class'''
        print("hobby is playing football")
```

```python
class child1(father1):
    def __init__(self,name,age,gender,department,salary,height,weight):
     super().__init__(name,age,gender,department,salary)
     self.height = height
     self.weight = weight

    def hobbies(self):
       print("hobby is playing hopscoth")
```

```python
obj2 = father1('a',60,'m','md',12345)
```
