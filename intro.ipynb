```python
#palindrome code
string = input("enter a string :  ")
string1 = string[::-1]

if string == string1:
    print('The String Is A Palindrome')
else:
    print('string is not a palindrome')
```

```python
#logical statement along with if else
''' 90 and abve -a+,80 to 90 -A ,70 - 80 B+,under 60  D '''
Marks = int(input('Enter Marks :'))
if (Marks > 90 and Marks == 90 ):
    print("Grade : A+")
elif (Marks >= 80 and Marks < 90):
    print("Grade : A")
elif (Marks >=70 and Marks < 80):
    print("Grade : B+")
elif (Marks >=60 and Marks <70):
    print("Grade : B")
else :
    print("Grade : D")

```

```python
# Concatenating two tuples

tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)

concatenated_tuple = tuple1 + tuple2
print(concatenated_tuple)  # Output: (1, 2, 3, 4, 5, 6)

# Repeating a tuple
repeated_tuple = tuple1 * 2
print(repeated_tuple)  # Output: (1, 2, 3, 1, 2, 3)
```

```python
# tuple Comprehension in Python

# Note: Python does not have a built-in syntax for tuple comprehensions like list comprehensions.
# However, you can create a tuple using a generator expression and the tuple() constructor.

squared_tuple = tuple(x**2 for x in range(5))
print(squared_tuple)  
```
