```python
from abc import ABC,abstractmethod

class bank(ABC):
    def __init__(self,userid,passcode,acc_no,
                 acc_holdername,age,location,gender):
        self.__userid = userid
        self.__passcode = passcode
        self.acc_no = acc_no
        self.acc_holdername =  acc_holdername
        self.age = age
        self.location = location
        self.gender =  gender

    @abstractmethod
    def credit_bal(self):
        pass

    @abstractmethod
    def debit_bal(self):
            pass

    @abstractmethod
    def transfer_bal(self):
            pass

    @abstractmethod
    def check_bal(self):
            pass

    
```

```python
class saving_acc(bank):
    def __init__(self,userid,passcode,acc_no,
                 acc_holdername,age,location,gender,interest = 0.8):
        super().__init__(userid,passcode,acc_no,
                 acc_holdername,age,location,gender)
        self.interest = interest
    def credit_bal(self):
        pass

    def debit_bal(self):
        pass
    
    
    def transfer_bal(self):
        pass
    
    
    def check_bal(self):
        pass

        
```
