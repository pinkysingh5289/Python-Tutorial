```python
#Write a program to calculate 3-point moving averages from a list.
# Type your code below
data = [10,20,30,40,50]
moving_averages = []
for i in range(len(data)-2):
    v = data[i]+data[i+1]+data[i+2]
    a = v/3
    moving_averages.append(a)
print(moving_averages)
```

```python
#Write a program to find the day(s) with the highest temperature from a list of tuples (day, temperature).
data = [('M',20),('T',35),('W',34)]
days = []
f=[]
for j,k in data :
    a = k
    f.append(a)
s = max(f)
for j,k in data :
    if k == s :
        days.append(j)
print(s)
print(days)
```

```python
# Type your code below Write a program to find the longest word in a given sentence.

sentence = "Python programming is fun"
a = sentence.split()
num = []
for i in a :
    count = 0
    for j in i :
        count = count + 1
    num.append(count)
new = list(zip(a,num))
m =[]
lists = []
for i,j in new:
    m.append(j)
large = max(m)
for i,j in new :
    if j==large:
        lists.append(i)
for i in lists:
    print(i)
        
```

```python
#Write a program to find the factorial of a number using recursion
# Type your code below
n = 5
def fact(n):
    if n == 0 or n ==1 :
        return 1
    return  n*fact(n-1)
a = fact(n)
print(a)
```
