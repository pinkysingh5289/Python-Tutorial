```python
# Program to print all prime numbers between 1 and N

N = 10   # You can change this value to test with other inputs

for num in range(2, N + 1):   # Start from 2 (smallest prime)
    is_prime = True
    for i in range(2, int(num**0.5) + 1):  # Check divisibility up to sqrt(num)
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num)
```
