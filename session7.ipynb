```python
class bank:
    bank_name = 'sbi'
    def __init__(self,acc_name,age,gender,location,balance,userid,passcode):
        self.acc_name = acc_name
        self.age = age
        self.gender = gender
        self.location = location
        self.balance = balance
        self.userid = userid # putting underscore 2 time - it will hide user_id(it also called encapsulate)
        self.passcode = passcode
    def credit_info(self):
        pass
    def debit_info(self):
        pass
```
