```python
a = 5
b = int(input("Enter : "))
try :
    result = a / b
    print("result", result)
except ZeroDivisionError as zde:
    print("error occur",zde)
except ValueError as ve:
    print("Error occur",ve)
except Exception as e :
    print(f"Exception occured  {e}")
else :
    print(f"This statement executed when no exception occurs")
finally :
    print(f"Done")

```

```python
import logging
logging.basicConfig(
    level =  logging.INFO,
    filename = 'model.log',
    filemode = 'w',
    format = '%(asctime)s - %(message)s - %(levelname)s',
    force =  True
)

logging.info("Logging setup successful")
try :
    # Code that might cause an error
    num = int(input("Enter a number: "))
    result = 10 / num
except ZeroDivisionError as ze :
    # Runs if division by zero happens
    print(f"error {ze}")
except ValueError as ve:
    # Runs if input is not a number
    print(f"valueerror as {ve}")
except Exception as e :
    # for any other type of error 
    print(f"error is {e}")
else :
    # Runs only if no error occurs
    print("result : ", result )
finally:
    # Always runs (cleanup code)
    print("Done")
```

```python
import logging
logging.basicConfig(
    level = logging.INFO,
    filename = 'model.log',
    filemode = 'w',
    format = '%(asctime)s - %(message)s - %(levelname)s',
    force = True
)

logging.info("logging setup ")
try :
    num1 = int(input("Enter a first number: "))
    num2 = int(input("Enter a second number: "))
    num3 = num1 / num2
    print("Result:", num3)
except ZeroDivisionError:
    logging.exception("Checkk value division by zero")
    print("Error is division by zero")
except ValueError:
    logging.exception("value error occur")
    print("value error,correct value input")
except TypeError:
    logging.exception("Check the values: Type error occurred")
    print("Error: Invalid type used in calculation!")
finally:
    print("done")
```
