```python
# Set Methods are as follows:
set = {1,2,3,8}
set2 = {3,4,5,6,2,1}
```

```python
value = set.union(set2)
print(value)
```

```python
value2 = set.intersection(set2)
print(value2)
```

```python
value3 = set.difference(set2)
print(value3)
```

```python
value4 = set2.difference(set)
print(value4)
```

```python
# Symmetric difference of two sets
value5 = set.symmetric_difference(set2)
print(value5)
```

```python
# Subset and Superset
print(set.issubset(value))
print(value.issuperset(set))
```

```python
def fact():
    num = int(input('Enter  a num : '))
    fact = 1
    for i in range(1,num+1):
        fact = fact * i
    print(f'factorial of {num} is {fact}')

fact()
```
