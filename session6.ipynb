```python
class bank:
    bank_name = 'SBI'

    def __init__(self,name,age,gender):
        self.name = name
        self.age =  age
        self.gender =  gender

    def credit_info(self,interest_rate = 8.0): #method definiton
        print(f'{self.name},your age is {self.age} and interest rate is {interest_rate}')

    def debit_info(self,interest_rate ):
        print(f'{self.age}')

name = input("Enter the name : ")
age = input("Enter the Age : ")
gender = input("Enter the Gender : ")

```

```python
object_a = bank(name,age,gender)
```

```python
object_a.credit_info() 
object_a.debit_info(8)
```

```python
#single inheritence 
class parent:
    def __init__(self):
        pass
    def method1(self):
        pass

#Child class 
class child1(parent):
    def __init__(self):
        pass
```

```python
#Mutli-Level /Multiple -Level inheritence : 
# it is not recommended to use because if there will be any issue in child 1 then child 2 and child 3  will also face problem and this will make website /application weak  
# and also it take more time to run -latency is higher
class parent:
    def __init__(self):
        pass
    def method1(self):
        pass

#Child class 
class child1(parent):
    def __init__(self):
        pass

class child2(child1):
    def __init__(self):
        pass


class child3(child2):
    def __init__(self):
        pass

```

```python
#Hybrid inhertience - It is a mixture of two classes(single and multi-level) 
class parent1:
    def __init__(self):
        pass
    def method1(self):
        pass

class parent2:
    def __init__(self):
        pass
    def method2(self):
        pass

class child(parent1,parent2):
    def __init__(self):
        pass
```

```python
#Herarchical inheritence -tree like structure-this is good method becuase of latency 

class parent:
    def __init__(self):
        pass
    def method1(self):
        pass

class child1(parent):
    def __init__(self):
        pass

class child2(parent):
    def __init__(self):
        pass

```
